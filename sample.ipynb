# Desafio DIO - ETL Com Python

Código em python para pegar dados de um arquivo csv, formatar todas as colunas igualmente e retorna em json

```python
import pandas as pd
import re
import json
```

Aqui iremos extrair os dados do csv para o pandas poder manipular

```python
df = pd.read_csv('data/Dio - desafio - Página1.csv')
df.head()
```

## Após ter os dados, vamos transforma-los com expressão regex mudando de endereço para: rua, número e bairro

```python
def separar_endereco(endereco):

    match = re.match(r'^(.*),\s*([^ -]+)\s*-\s*(.*)$', endereco)
    if match:
        return pd.Series([match.group(1).strip(), match.group(2).strip(), match.group(3).strip()])
    else:
        return pd.Series([None, None, None])
```

## Transforma o telefone no padrão desejado de (DDD) XXXXX-XXXX

```python
def formatar_telefone(telefone):
    limpo = re.sub(r'\D', '', str(telefone))


    if len(limpo) == 11:
        return f"({limpo[:2]}) {limpo[2:7]}-{limpo[7:]}"

    elif len(limpo) == 9:
        return f"(19) {limpo[:5]}-{limpo[5:]}"

    return telefone
```

## Chamar as funções e criar o novo DataFrame e um arquivo json com os dados

```python
df[['rua', 'numero', 'bairro']] = df['endereco'].apply(separar_endereco)
df['telefone'] = df['telefone'].apply(formatar_telefone)

df_final = df[['nome', 'rua', 'numero', 'bairro', 'telefone']]


json_resultado = df_final.to_json(orient='records', force_ascii=False, indent=4)

print(json_resultado)
```
